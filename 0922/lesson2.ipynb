```python
#1 到 100 的奇數與偶數加總
even_total = 0
odd_total = 0
for i in range(1, 101):
	if i % 2 == 0:
		even_total += i
	else:
		odd_total += i
print(f"1加到100偶數的加總為 {even_total}")
print(f"1加到100奇數的加總為 {odd_total}")
```
