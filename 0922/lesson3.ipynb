```python
#1個學生5個分數
chinese = 78
english = 85
math = 85
history = 92
discovery = 100

total = chinese + english + math + history + discovery
average = total / 5

print(f"學生總分{total},平均{average:.2f}")
```

```python
#list資料結構
scores = [
	78,
  85,
  85,
  92,
 100
]
total = 0
for score in scores:
	total += score

average = total / 5
print(f"學生總分{total},平均{average:.2f}")
```

```python
import json
from urllib.request import Request, urlopen

url = "https://tcgbusfs.blob.core.windows.net/dotapp/youbike/v2/youbike_immediate.json"
request = Request(url, headers={"User-Agent": "Mozilla/5.0"})

with urlopen(request, timeout=15) as response:
    youbike_data = json.load(response)

print(f"共取得 {len(youbike_data)} 筆資料")
youbike_data[:3]
```
