```python
import math

# 初始設定
principal = 20000
target = 22000
rate = 0.011

# 1 + r
growth_factor = 1 + rate

# 方程式簡化：growth_factor^n = target_ratio
target_ratio = target / principal

# 利用對數求解：n = log(target_ratio) / log(growth_factor)
n = math.log(target_ratio) / math.log(growth_factor)

print(f"存款從 {principal:,} 元增長到 {target:,} 元（年利率 {rate*100}% 複利）")
print(f"需要的時間大約為：{n:.2f} 年")
```
