### 什麼是 `float()`？

`float()` 函數可以將傳入的資料轉換成「浮點數」（帶有小數點的數字）。它最常用於以下幾種情境：

1. **將字串（str）轉換為浮點數**：
   當我們使用 `input()` 讓使用者輸入數值時，Python 預設會將輸入內容當成「字串」（如 `"27.156"`）。若要進行數學運算，必須先用 `float()` 將它轉換為數值。
2. **將整數（int）轉換為浮點數**：
   將整數轉換為浮點數後，數值會自動加上 `.0`。

下面我們來看幾種常見的 `float()` 使用範例：

```python
# 1. 將整數轉換成浮點數
num_int = 15
num_float1 = float(num_int)
print("整數轉浮點數:", num_float1)  # 輸出 15.0

# 2. 將包含數字的字串轉換成浮點數
str_num = "3.14159"
num_float2 = float(str_num)
print("字串轉浮點數:", num_float2)  # 輸出 3.14159

# 3. 搭配 input() 接收使用者輸入（如您的上一個儲存格）
# 如果使用者輸入 10，float("10") 會成功轉換為 10.0
user_input = "10"
print("輸入轉浮點數:", float(user_input))

# 4. 常見錯誤提示：如果字串內含有非數字字元，轉換會失敗並產生 ValueError
try:
    float("apple")
except ValueError as e:
    print("\n錯誤示範 (無法將非數字字串轉換成 float):", e)
```

### 實作練習：計算圓柱體體積

圓柱體體積公式：
$$\text{體積} = \pi \times r^2 \times h$$
其中：
* $r$ 為圓柱底面的**半徑**
* $h$ 為圓柱的**高**
* $\pi$ 為圓周率（我們可以使用 Python 內建 `math` 模組中的 `math.pi`，大約是 3.14159）

```python
import math

# 1. 取得使用者輸入並使用 float() 轉換
radius = float(input("請輸入圓柱體的底面半徑 (r): "))
height = float(input("請輸入圓柱體的高 (h): "))

# 2. 計算圓柱體體積
# math.pi 代表圓周率，radius ** 2 代表半徑的平方
volume = math.pi * (radius ** 2) * height

# 3. 顯示計算結果
print("===========================")
print(f"圓柱體半徑: {radius}")
print(f"圓柱體高度: {height}")
print(f"計算出來的圓柱體體積為: {volume:.2f}")  # :.2f 代表四捨五入保留到小數點後兩位
```
